```python
from sklearn.compose import ColumnTransformer
```

```python
from sklearn.datasets import load_iris
```

```python
iris=load_iris();
```

```python
import pandas as pd
```

```python
df=pd.DataFrame(iris.data,columns=iris.feature_names)
```

```python
df
```

```python
import matplotlib.pyplot as plt
df['sepal length (cm)'].plot(kind='kde')
df['sepal width (cm)'].plot(kind='kde')
df['petal length (cm)'].plot(kind='kde')
df['petal width (cm)'].plot(kind='kde')
plt.legend()
```

```python
s=['sepal length (cm)','sepal width (cm)']
m=['petal length (cm)','petal width (cm)']
```

```python
transformer=ColumnTransformer([('st',StandardScaler(),s),('mt',MinMaxScaler(),m)],remainder='passthrough')
```

```python
transformed_df=transformer.fit_transform(df)
```

```python
import numpy as np
transformed_df=pd.DataFrame(transformed_df,columns=df.columns)
```

```python
transformed_df
```

```python
transformed_df['sepal length (cm)'].plot(kind='kde')
transformed_df['sepal width (cm)'].plot(kind='kde')
transformed_df['petal length (cm)'].plot(kind='kde')
transformed_df['petal width (cm)'].plot(kind='kde')
plt.legend()
```
