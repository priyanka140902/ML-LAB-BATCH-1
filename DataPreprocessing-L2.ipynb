```python
import seaborn as sns
l=sns.get_dataset_names()
```

```python
l
```

```python
df=sns.load_dataset('diamonds')
df
```

```python
import matplotlib.pyplot as plt
plt.bar(df['clarity'].value_counts().index,df['clarity'].value_counts().values, color='blue',edgecolor='yellow')
plt.xlabel('clarity')
plt.ylabel('frequency')
plt.show()
```

```python
plt.hist(df['price'].values,color='red',histtype='stepfilled',edgecolor='green')
plt.xlabel('Price')
plt.show()
```

```python
plt.scatter(df['carat'],df['price'],color='green')
```

```python
df.corr(numeric_only=True)
```

```python
plt.pie(df['clarity'].value_counts().values,labels=df['clarity'].value_counts().index)
```
