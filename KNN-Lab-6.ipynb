```python
import numpy as np
from collections import Counter

class WeightedKNN:
    def __init__(self, k=3):
        self.k = k
    
    def fit(self, X_train, y_train):
        self.X_train = np.array(X_train)
        self.y_train = np.array(y_train)
    
    def predict(self, X_test):
        predictions = []
        for x in X_test:
            # Compute Euclidean distance
            distances = np.linalg.norm(self.X_train - x, axis=1)
            
            # Get indices of k nearest neighbors
            nn_indices = np.argsort(distances)[:self.k]
            
            # Compute weights = 1 / distance (avoid div by zero)
            weights = 1 / (distances[nn_indices] + 1e-8)
            
            # Weighted votes
            class_weights = {}
            for idx, w in zip(nn_indices, weights):
                label = self.y_train[idx]
                class_weights[label] = class_weights.get(label, 0) + w
            
            # Choose class with max weight
            predicted_class = max(class_weights, key=class_weights.get)
            predictions.append(predicted_class)
        
        return predictions


# Example usage
if __name__ == "__main__":
    # Training data (X, y)
    X_train = [[1, 2], [2, 3], [3, 3], [6, 7], [7, 8], [8, 9]]
    y_train = ["A", "A", "A", "B", "B", "B"]

    # Test data
    X_test = [[2, 2], [6, 6]]

    model = WeightedKNN(k=3)
    model.fit(X_train, y_train)
    predictions = model.predict(X_test)

    print("Predictions:", predictions)
```
